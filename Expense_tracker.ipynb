```python
expense = []
```

```python
expense_1 = {
    "amount": 12.50,
    "category": "Food",
    "description": "Lunch",
    "date": "2026-10-09"
}
expense_2 = {
    "amount": 82.90,
    "category": "Electronics",
    "description": "Watch",
    "date": "2026-10-09"
}

expense.append(expense_1)
print(expense_1)
expense.append(expense_2)
print(expense_2)


```

```python
for x in expense:
    print(x["category"],x["amount"],x["description"])
    
```

```python
total = 0

for x in expense:
    total += x["amount"]

print(f"Total spent: £{total:.2f}")
```
